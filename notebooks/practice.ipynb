### Generación de datos
Se generan dos distribuciones aleatorias de 1.000 observaciones cada una: una distribución normal con media 50 y desvío estándar 10, y una distribución uniforme en el rango de 20 a 80. Ambas se centran en 50 para poder compararlas.

```python
import matplotlib.pyplot as plt
import numpy as np

normal = np.random.normal(loc=50, scale=10, size=1000)
uniforme = np.random.uniform(low=20, high=80, size=1000)

print(normal[:5])
print(uniforme[:5])
```

### Análisis estadístico
Se calculan los estadísticos descriptivos de cada distribución: media, mediana, varianza, desvío estándar, mínimo y máximo.

```python
print('Distribución normal')
print('Media:', np.mean(normal))
print('Mediana:', np.median(normal))
print('Varianza:', np.var(normal))
print('Desvío estándar:', np.std(normal))
print('Mínimo:', np.min(normal))
print('Máximo:', np.max(normal))

print()

print('Distribución uniforme')
print('Media:', np.mean(uniforme))
print('Mediana:', np.median(uniforme))
print('Varianza:', np.var(uniforme))
print('Desvío estándar:', np.std(uniforme))
print('Mínimo:', np.min(uniforme))
print('Máximo:', np.max(uniforme))
```

### Visualización
Se crea una figura con dos subplots: a la izquierda el histograma de la distribución normal y a la derecha el de la distribución uniforme. Cada gráfico incluye título, etiquetas de ejes, leyenda y grilla, y la figura tiene un título general.

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 5))

ax[0].hist(normal, bins='sqrt', color='steelblue', edgecolor='white',
           label='Normal (media 50, desvío 10)')
ax[0].set_title('Distribución normal')
ax[0].set_xlabel('Valor')
ax[0].set_ylabel('Frecuencia')
ax[0].legend()
ax[0].grid(alpha=0.3)

ax[1].hist(uniforme, bins='sqrt', color='darkorange', edgecolor='white',
           label='Uniforme (rango 20 a 80)')
ax[1].set_title('Distribución uniforme')
ax[1].set_xlabel('Valor')
ax[1].set_ylabel('Frecuencia')
ax[1].legend()
ax[1].grid(alpha=0.3)

plt.suptitle('Comparación de distribuciones: normal y uniforme '
             '(1.000 observaciones)')
plt.tight_layout()
plt.show()
```

### Reflexión
Ambas distribuciones tienen una media cercana a 50, pero su forma es muy distinta.

En la distribución normal, los valores se concentran alrededor de la media y se vuelven menos frecuentes a medida que se alejan de ella, lo que genera la forma de campana del histograma. La media y la mediana son casi iguales, lo que indica que la distribución es simétrica. El desvío estándar, de aproximadamente 10, muestra que la mayoría de los datos se encuentra entre 40 y 60. Además, pueden aparecer valores extremos poco frecuentes alejados del centro.

En la distribución uniforme, todos los valores dentro del rango de 20 a 80 tienen la misma probabilidad de aparecer, por lo que el histograma presenta barras de altura similar en todo el rango. Su varianza y su desvío estándar (aproximadamente 300 y 17) son mayores que los de la normal, porque los datos están repartidos de forma pareja y no concentrados en el centro. Los valores mínimo y máximo quedan muy cerca de los límites definidos (20 y 80), y nunca los superan.

En síntesis, la distribución normal concentra los datos en torno a la media, mientras que la uniforme los distribuye de forma homogénea dentro de su rango.
